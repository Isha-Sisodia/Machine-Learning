### Step 1: Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score   #for checking Accuracy
```

### Step 2: Load dataset

```python
df = pd.read_csv("Salary_Data.csv")
print(df.head())
```

### Step 3: Define X and Y

```python
X = df[["YearsExperience"]]
Y = df["Salary"]
```

### Step 4: Train-Test Split

```python
# 👉 Use 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)
```

### Step 5: Train the model

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

### Step 6: Make predictions

```python
y_pred = model.predict(X_test)
```

### Step 7: Plot graph

```python
plt.scatter(X, Y)  # actual data
plt.plot(X_train, model.predict(X_train))  # regression line
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Salary vs Experience")
plt.show()
```

### Step 8: Check accuracy

```python
print("Accuracy:", r2_score(y_test, y_pred))
```

### Step 9: Predict your own value

```python
exp = np.array([[5]])   # 5 years experience
predicted_salary = model.predict(exp)

print("Predicted Salary:", predicted_salary)
```

✅ FIX (Best Practice)

👉 Instead of NumPy array, use DataFrame

```python
exp = pd.DataFrame([[5]], columns=['YearsExperience'])

predicted_salary = model.predict(exp)

print("Predicted Salary:", predicted_salary)
```

```python
print("Slope (m):", model.coef_)
print("Intercept (b):", model.intercept_)
```

```python
comparison = pd.DataFrame({
    "Actual Salary": y_test,
    "Predicted Salary": y_pred
})

print(comparison)
```

```python
comparison["Error"] = comparison["Actual Salary"] - comparison["Predicted Salary"]
print(comparison)
```

```python
plt.scatter(X_train, y_train, color='blue', label="Training Data")

plt.scatter(X_test, y_test, color='orange', label="Test Data (Actual)")

plt.scatter(X_test, y_pred, color='green', label="Test Data (Predicted)")

plt.plot(X_train, model.predict(X_train), color='red', label="Regression Line")

plt.xlabel("Experience")
plt.ylabel("Salary")
plt.legend()
plt.show()
```

### 📈 What this graph shows
🔵 Training data → used to learn

🟠 Test data (actual) → real unseen values

🟢 Test predictions → model’s guess

🔴 Line → learned relationship
